```python
duplicate_rows = dnkn.duplicated().sum()
print(duplicate_rows)
```

```python
duplicate_records = dnkn[dnkn.duplicated()]
print(duplicate_records)
```

```python
columns_to_check = ['Total Fat (g)', 'Saturated Fat (g)', 'Trans Fat (g)', 'Cholesterol (mg)', 'Sodium (mg)', 'Fiber (g)', 'Protein (g)', 'Carbs (g)']
for col in columns_to_check:
    if col in dnkn.columns:
        print(f"\nUnique values in '{col}':")
        print(dnkn[col].unique()[:50])
else:
    print(f"\nColumn '{col}' not found in the DataFrame.")
```

```python
dnkn1 = dnkn.copy()
print("DataFrame 'dnkn1' created as a copy of 'dnkn'.")
```

```python
food_items = ['Ham and Cheese Rollups', 'Coffee', 'Espresso Drinks', 'Smoothies', 'Iced Coffee', 'Frappés', 'Hot Coffee', 'Hot Chocolate', 'Tea', 'Iced Tea', 'Lemonade', 'Frozen Coffee', 'Frozen Lemonade', 'Frozen Smoothie', 'Frozen Tea', 'Frozen Frappé', 'Frozen Hot Chocolate', 'Frozen Espresso Drink', 'Frozen Iced Coffee', 'Frozen Iced Tea', 'Frozen Lemonade', 'Frozen Smoothie', 'Frozen Tea', 'Frozen Frappé', 'Frozen Hot Chocolate', 'Frozen Espresso Drink', 'Frozen Iced Coffee', 'Frozen Iced Tea', 'Frozen Lemonade', 'Frozen Smoothie', 'Frozen Tea', 'Frozen Frappé', 'Frozen Hot Chocolate', 'Frozen Espresso Drink', 'Frozen Iced Coffee', 'Frozen Iced Tea', 'Frozen Lemonade', 'Frozen Smoothie', 'Frozen Tea', 'Frozen Frappé', 'Frozen Hot Chocolate', 'Frozen Espresso Drink', 'Frozen Iced Coffee', 'Frozen Iced Tea', 'Frozen Lemonade', 'Frozen Smoothie', 'Frozen Tea', 'Frozen Frappé', 'Frozen Hot Chocolate']
```

```python
dnkn1 = dnkn1[~dnkn1['Category'].isin(food_items)].copy()
print("Food items removed from the 'Category' column.")
```

```python
duplicate_check_after_removal = dnkn1.duplicated().sum()
print(f"Number of duplicate records after removing food items: {duplicate_check_after_removal}")
```

```python
dnkn2 = dnkn1.copy()
```

```python
dnkn2['Cholesterol (mg)'] = dnkn2['Cholesterol (mg)'].replace('<5', '5')
print(" '<5' replaced with '5' in 'Cholesterol (mg)' column.")
```

```python
for col in dnkn2.select_dtypes(include='object').columns:
    dnkn2[col] = dnkn2[col].str.replace('\xa0', '', regex=False).str.strip()
print(" '\xa0' removed and whitespace stripped from object columns.")
```

```python
for col in dnkn2.columns:
    if dnkn2[col].dtype == 'object':
        print('\n', col)
        print(dnkn2[col].unique()[:50])
```

```python
import numpy as np
```

```python
dnkn2 = dnkn2.replace('', np.nan)
print("Empty strings replaced with NaN values.")
```

```python
print(dnkn2.isnull().sum())
```

```python
dnkn3 = dnkn2.dropna().copy()
print("NaN values dropped and 'dnkn3' DataFrame created.")
```

```python
print(dnkn3.isnull().sum())
```

```python
numeric_columns = ['Total Fat (g)', 'Saturated Fat (g)', 'Trans Fat (g)', 'Cholesterol (mg)', 'Sodium (mg)', 'Fiber (g)', 'Protein (g)', 'Carbs (g)']
```

```python
for col in numeric_columns:
    dnkn3[col] = pd.to_numeric(dnkn3[col])
```

```python
print(" MISSING VALUES ")
print(dnkn3.isnull().sum())

print("\n DATA TYPES ")
print(dnkn3.dtypes)

print("\n DUPLICATE RECORDS ")
print('Duplicates:', dnkn3.duplicated().sum())

print("\n CATEGORY CHECK ")
print(dnkn3['Category'].value_counts())

print("\n SPECIAL VALUE CHECK ")
print("Cholesterol Values:")
print(dnkn3['Cholesterol (mg)'].unique())
print("\nTotal Fat Values:")
print(dnkn3['Total Fat (g)'].unique())

print("\n DATASET SHAPE ")
print(dnkn3.shape)

print("\n SUMMARY STATISTICS ")
print(dnkn3.describe().round(2))
```

```python
dnkn3.to_csv('dnkn_clean.csv', index=False)
```

```python
In Deliverable Two, the primary objective was to clean and prepare the Dunkin' Donuts nutritional dataset for analysis. Initially, the dataset contained several issues that could impede accurate statistical interpretation. These included the presence of duplicate records, which could lead to overcounting observations and special characters within numerical columns, such as "<5" and "\xa0" that prevented proper data type conversion and numerical operations. Additionally, some rows represented food categories rather than specific beverage products and there were empty string entries that needed to be treated as missing values. To address these issues, a systematic cleaning process was undertaken. Duplicate records were identified and removed. Special characters were systematically cleaned from relevant columns and empty strings were converted into NaN (Not a Number) values to be properly handled as missing data. Rows identified as food categories were filtered out to ensure that the dataset exclusively contained beverage items. Finally, all relevant columns were converted to appropriate numeric data types, and the entire dataset was validated through summary statistics, data type checks, and shape verification to ensure it was clean, consistent, and ready for further analysis.
```
